## Задание


Необходимо вычислить значение выражения:
$$H=\frac{g \cdot h}{d17} + b/h1 - \frac{x + y}{4}$$
Для каждой переменной необходимо вывести ее значение и тип.

## Анализ выражения и определение переменных

Проанализируем выражение и определим переменные, которые потребуются для выполнения задания. Независимые переменные: $g, h, d17, b, h1, x, y$. Переменная, которая должна быть вычислена на основе независимых: $H$.

```python
g = 5
h = 20
d17 = 1.1
b = 20
h1 = 10
x = 2
y = 4
print("g = ", g, ", type(g) = ", type(g))
print("h = ", h, ", type(h) = ", type(h))
print("d17 = ", d17, ", type(d17) = ", type(d17))
print("b = ", b, ", type(b) = ", type(b))
print("h1 = ", h1, ", type(h1) = ", type(h1))
print("x = ", x, ", type(x) = ", type(x))
print("y = ", y, ", type(y) = ", type(y))
```

## Вычисление выражение, печать результатов

Код для вычисления значения выражения $H$:

```python
H = ((g * h) / d17) + b / h1 - (x + y) / 4
print('H=',H)
print('Type of H: ', type(H))
```

Основные функции, используемые в данной практической работе: `print(), type(), abs()`
